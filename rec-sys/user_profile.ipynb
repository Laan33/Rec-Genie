```python

import preprocessing as pre
import pandas as pd
import data_loader
import user_profile as user_pf
```

```python
# Load the data
films_df = data_loader.load_movies()
ratings_df = data_loader.load_ratings()
credits_df = data_loader.load_credits()
```

```python
films_df.head()
```

```python

print("Data dimensions:")
print("films_df: ", films_df.shape)
print("ratings_df: ", ratings_df.shape)
print("credits_df: ", credits_df.shape)
```

```python
ohe_films_df, genre_list_mlb = pre.one_hot_encode_genres(films_df)
```

```python
films_df = pre.data_tidying(ohe_films_df, credits_df)
print(films_df.columns)
```

```python
# remove all films released before 1985
films_df = films_df[pd.to_datetime(films_df['release_date']).dt.year >= 1985]
```

```python
films_df.sort_values(by='release_date', inplace=True, ascending=False)
films_df.head()
```

```python
# Find common movieIds
common_movie_ids = (set(films_df['id'])
                    # .intersection(set(ratings_df['movieId']))
                    .intersection(set(credits_df['id'])))

# Filter datasets to include only common movieIds
filtered_films_df = films_df[films_df['id'].isin(common_movie_ids)]
filtered_ratings_df = ratings_df[ratings_df['movieId'].isin(common_movie_ids)]
filtered_credits_df = credits_df[credits_df['id'].isin(common_movie_ids)]

# Count the number of entries in each filtered dataset
num_films = len(filtered_films_df)
num_ratings = len(filtered_ratings_df)
num_credits = len(filtered_credits_df)

print(f"Number of films: {num_films}")
print(f"Number of ratings: {num_ratings}")
print(f"Number of credits: {num_credits}")
```
